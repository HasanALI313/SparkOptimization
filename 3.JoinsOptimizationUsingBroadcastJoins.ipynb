Join OPimizationUsing Broadcast Joins
- Joins are usually the most expensive more so when distrbutive environment.
- let's say we have Fact table of 1GB and DIM table of 1MB
- Base on the 128MB Rule there will 8 partitions for the Fact Table and 1 partition for the DIM table.

```python
from pyspark.sql.functions import *
```

```python
stadiums_dict = {'Anfield': 'Liverpool', 'Goodison Park': 'Everton', 'Emirates Stadium': 'Arsenal', 'Stamford Bridge': 'Chelsea', 'Tottenham Hotspur Stadium': 'Tottenham Hotspur', 'St Mary’s Stadium': 'Southampton', 'Etihad Stadium': 'Manchester City', 'Old Trafford': 'Manchester United', 'Villa Park': 'Aston Villa', 'St. James’ Park': 'Newcastle United', 'Bramall Lane': 'Sheffield United', 'St. Andrew’s': 'Stoke City', 'Vetch Field': 'Bolton Wanderers', 'The Hawthorns': 'Birmingham City', 'The Den': 'Brentford', 'The Valley': 'Bristol City', 'The Hawthorns': 'Bristol Rovers'}
stadium_df = spark.createDataFrame(stadiums_dict.items(), ['Stadium', 'Team'])

epl_df = spark.read.table('sparkoptimization.tables.epl_final')
```

```python
joined_df = epl_df.join(stadium_df, epl_df.HomeTeam == stadium_df.Team, 'left')
display(joined_df)
```

#Broadcaser the join

```python
from pyspark.sql.functions import broadcast

joined_df = epl_df.join(broadcast(stadium_df), epl_df.HomeTeam == stadium_df.Team, 'left')
display(joined_df)
```
