# Adaptive Query Execution (AQE)
-------------------------------------
Adaptive Query Execution (AQE) is one of Spark’s most important performance features.

It means Spark can change the query plan while the query is running, based on real runtime data instead of guesses made at compile time.

1. Usually spark will
- Plan → Execute → Finish

2. With AQE Spark will
- Plan → Start executing → Collect runtime stats → RE-OPTIMISE → Continue 

⚙️ Why AQE exists

Before AQE, Spark relied on:

table statistics (often outdated)
estimated data sizes
static join decisions

This often caused:

bad join choices
huge shuffles
skew issues
inefficient partitions

AQE fixes this by using real execution data.

Stage 1 → collect stats
Stage 2 → re-optimize plan
Stage 3 → continue execution

- i.e. 
Spark Initially chooses to SortMergeJoin , at runtime it finds one table is very small so it switches to a BroadCastHashJoin, is faster query and no restart is required.

| Feature                | What it does                         |
| ---------------------- | ------------------------------------ |
| Dynamic Join Selection | switches join types during execution |
| Skew Join Handling     | splits heavy partitions              |
| Coalesce Partitions    | reduces unnecessary partitions       |
| Runtime Statistics     | uses real data instead of estimates  |
